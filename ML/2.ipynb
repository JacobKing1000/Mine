```python
import pandas as pd
import matplotlib.pyplot as plt

Airbnb_df = pd.read_csv(""C:\Users\chand\Downloads\Airbnb_data.xlsx"")

print(Airbnb_df.head())
print(Airbnb_df.info())
print(Airbnb_df.duplicated().sum())
```

```python
# Graphs

# Histogram
Airbnb_df.hist(figsize=(12,10))
plt.suptitle("Histogram of Numerical Columns")
plt.show()

# Box Plot
num_cols = Airbnb_df.select_dtypes(include='number').columns
Airbnb_df[num_cols].plot(kind='box', subplots=True, layout=(len(num_cols), 1), figsize=(8, 3*len(num_cols)))
plt.suptitle("Box Plot")
plt.show()

# Bar Chart
cat_cols = Airbnb_df.select_dtypes(include='object').columns

if len(cat_cols) > 0:
    Airbnb_df[cat_cols[0]].value_counts().head(10).plot(kind='bar', figsize=(8,5))
    plt.title(f"Top 10 Values of {cat_cols[0]}")
    plt.xlabel(cat_cols[0])
    plt.ylabel("Count")
    plt.show()

# Pie Chart
if len(cat_cols) > 0:
    Airbnb_df[cat_cols[0]].value_counts().head(5).plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
    plt.ylabel("")
    plt.title(f"Distribution of {cat_cols[0]}")
    plt.show()

corr = Airbnb_df[num_cols].corr()

plt.figure(figsize=(10,8))
plt.imshow(corr, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
