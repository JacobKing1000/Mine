```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
print(a)

b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b)

print(a + 2)
print(a - 2)
print(a * 2)
print(a / 2)
print(a ** 2)

print(b.shape)
print(b.ndim)
print(b.size)
print(b.dtype)

print(a[0])
print(a[-1])
print(a[1:4])

print(b[1, 2])
print(b[0])
print(b[:, 1])

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print(A)
print(B)

print(A + B)
print(A - B)
print(A * B)
print(A @ B)
print(A.T)

print(np.sum(a))
print(np.mean(a))
print(np.max(a))
print(np.min(a))
```
